```python
import pandas as pd
import numpy as np
import plotly.graph_objects as go
from plotly.subplots import make_subplots

df["date"] = pd.to_datetime(df["date"])

monthly = (
    df.groupby("year_month")
      .agg(
          new_revenue=("new_customer_revenue", "sum"),
          repeat_revenue=("repeat_customer_revenue", "sum"),
          total_revenue=("total_revenue", "sum")
      )
      .reset_index()
)
monthly["date"] = pd.to_datetime(monthly["year_month"])

spend_cols = [c for c in df.columns if c.startswith("spend_") or c == "Tiktok_DTC_Spends"]

channel_names = {
    "spend_Meta_Asc_Retargeting": "Meta ASC Retargeting",
    "spend_Meta_Asc": "Meta ASC",
    "spend_Meta_Retargeting_Manual": "Meta Retargeting",
    "spend_Meta_Broad_Manual": "Meta Broad",
    "spend_Meta_Non_Sales": "Meta Non-Sales",
    "spend_google_PMAX_Exist": "Google PMAX Existing",
    "spend_google_Search_Non_Brand": "Google Search Non-Brand",
    "spend_google_PMAX_New": "Google PMAX New",
    "spend_google_Search_Brand": "Google Search Brand",
    "spend_google_Shopping": "Google Shopping",
    "spend_snapchat": "Snapchat",
    "spend_Youtube": "YouTube",
    "Tiktok_DTC_Spends": "TikTok"
}

channel_spend = df[spend_cols].sum().sort_values()
channel_spend.index = [channel_names.get(x, x) for x in channel_spend.index]
channel_share = channel_spend / channel_spend.sum() * 100

total_revenue = df["total_revenue"].sum()
total_spend = df["total_media_spend"].sum()
revenue_per_spend = total_revenue / total_spend
new_share = df["new_customer_revenue"].sum() / total_revenue

bfcm = df[df["BFCM_promo"] == 1]
non_bfcm = df[df["BFCM_promo"] == 0]
bfcm_avg = bfcm["total_revenue"].mean()
non_bfcm_avg = non_bfcm["total_revenue"].mean()

df["week_of_year"] = df["date"].dt.isocalendar().week.astype(int)
d2025 = df[(df["year"] == 2025) & (df["week_of_year"] <= 34)]
d2026 = df[(df["year"] == 2026) & (df["week_of_year"] <= 34)]

rev25, rev26 = d2025["total_revenue"].sum(), d2026["total_revenue"].sum()
spend25, spend26 = d2025["total_media_spend"].sum(), d2026["total_media_spend"].sum()
rev_change = (rev26 / rev25 - 1) * 100
spend_change = (spend26 / spend25 - 1) * 100
eff25, eff26 = rev25 / spend25, rev26 / spend26

peak = df.loc[df["total_revenue"].idxmax()]

BLUE = "#2E5EAA"
GREEN = "#4FAE7C"
ORANGE = "#E4572E"
TEXT_DARK = "#1A1A1A"
TEXT_MUTED = "#666666"
GRID = "rgba(0,0,0,0.07)"
CARD_BG = "#F5F7FA"

fig = make_subplots(
    rows=4, cols=4,
    specs=[
        [{"colspan": 4}, None, None, None],
        [{"colspan": 2}, None, {"colspan": 2}, None],
        [{"colspan": 2}, None, {"colspan": 2}, None],
        [{"colspan": 4}, None, None, None]
    ],
    row_heights=[0.04, 0.35, 0.35, 0.15],
    vertical_spacing=0.15,
    horizontal_spacing=0.09,
    subplot_titles=(
        "", "Revenue Trend", "Monthly Revenue by Customer Type",
        "Where Media Budget Is Going", "Revenue vs Media Spend"
    )
)

fig.update_xaxes(visible=False, row=1, col=1)
fig.update_yaxes(visible=False, row=1, col=1)

fig.add_trace(
    go.Scatter(
        x=df["date"], y=df["total_revenue"] / 1e6, mode="lines",
        name="Revenue", line=dict(width=2.5, color=BLUE),
        fill="tozeroy", fillcolor="rgba(46,94,170,0.08)",
        hovertemplate="<b>%{x|%d %b %Y}</b><br>Revenue: $%{y:.2f}M<extra></extra>"
    ),
    row=2, col=1
)
fig.add_trace(
    go.Scatter(
        x=bfcm["date"], y=bfcm["total_revenue"] / 1e6, mode="markers",
        name="BFCM weeks",
        marker=dict(size=9, symbol="circle-open", color=ORANGE, line=dict(width=2)),
        hovertemplate="<b>BFCM Week</b><br>%{x|%d %b %Y}<br>Revenue: $%{y:.2f}M<extra></extra>"
    ),
    row=2, col=1
)
fig.add_annotation(
    x=peak["date"], y=peak["total_revenue"] / 1e6, xref="x", yref="y",
    text=f"<b>Peak<br>${peak['total_revenue']/1e6:.2f}M</b>",
    showarrow=True, arrowhead=2, arrowcolor=TEXT_MUTED, ax=0, ay=-40,
    font=dict(size=11, color=TEXT_DARK)
)

fig.add_trace(
    go.Scatter(
        x=monthly["date"], y=monthly["new_revenue"] / 1e6, mode="lines",
        name="New Customers", line=dict(width=2.5, color=BLUE),
        hovertemplate="<b>%{x|%b %Y}</b><br>New Revenue: $%{y:.2f}M<extra></extra>"
    ),
    row=2, col=3
)
fig.add_trace(
    go.Scatter(
        x=monthly["date"], y=monthly["repeat_revenue"] / 1e6, mode="lines",
        name="Repeat Customers", line=dict(width=2.5, color=GREEN),
        hovertemplate="<b>%{x|%b %Y}</b><br>Repeat Revenue: $%{y:.2f}M<extra></extra>"
    ),
    row=2, col=3
)

fig.add_trace(
    go.Bar(
        x=channel_spend.values / 1e6, y=channel_spend.index, orientation="h",
        name="Media Spend", showlegend=False, marker_color=BLUE,
        customdata=channel_share.values,
        hovertemplate="<b>%{y}</b><br>Spend: $%{x:.2f}M<br>Share: %{customdata:.1f}%<extra></extra>"
    ),
    row=3, col=1
)

fig.add_trace(
    go.Scatter(
        x=df["total_media_spend"] / 1000, y=df["total_revenue"] / 1e6,
        mode="markers", name="Weekly data",
        marker=dict(size=7, opacity=0.7, color=ORANGE),
        hovertemplate="<b>%{x:.0f}K spend</b><br>Revenue: $%{y:.2f}M<extra></extra>"
    ),
    row=3, col=3
)

kpis = [
    (f"${total_revenue/1e6:.2f}M", "TOTAL REVENUE"),
    (f"${total_spend/1e6:.2f}M", "MEDIA SPEND"),
    (f"{revenue_per_spend:.2f}x", "REVENUE / MEDIA SPEND"),
    (f"{new_share:.1%}", "NEW CUSTOMER SHARE")
]
positions = [(0.01, 0.25), (0.26, 0.50), (0.51, 0.75), (0.76, 0.99)]

for (value, label), (x0, x1) in zip(kpis, positions):
    fig.add_shape(
        type="rect", xref="paper", yref="paper",
        x0=x0, x1=x1, y0=0.965, y1=1.0,
        line=dict(width=1, color="rgba(0,0,0,0.06)"), fillcolor=CARD_BG
    )
    fig.add_annotation(
        x=(x0 + x1) / 2, y=0.99, xref="paper", yref="paper",
        text=f"<b>{value}</b>", showarrow=False, font=dict(size=16, color=TEXT_DARK)
    )
    fig.add_annotation(
        x=(x0 + x1) / 2, y=0.972, xref="paper", yref="paper",
        text=label, showarrow=False, font=dict(size=8, color=TEXT_MUTED)
    )

fig.add_annotation(
    x=0.02, y=0.065, xref="paper", yref="paper",
    text=f"<b style='color:{BLUE}'>BFCM PERFORMANCE</b><br>${bfcm_avg/1e6:.2f}M average weekly revenue during BFCM vs ${non_bfcm_avg/1e6:.2f}M otherwise",
    showarrow=False, xanchor="left", align="left", font=dict(size=11, color=TEXT_DARK)
)
fig.add_annotation(
    x=0.36, y=0.065, xref="paper", yref="paper",
    text=f"<b style='color:{GREEN}'>2026 YTD</b><br>Revenue {rev_change:+.1f}% vs 2025 · Media spend {spend_change:+.1f}%",
    showarrow=False, xanchor="left", align="left", font=dict(size=11, color=TEXT_DARK)
)
fig.add_annotation(
    x=0.68, y=0.065, xref="paper", yref="paper",
    text=f"<b style='color:{ORANGE}'>MEDIA EFFICIENCY</b><br>{eff25:.2f}x → {eff26:.2f}x revenue/media spend (first 34 weeks)",
    showarrow=False, xanchor="left", align="left", font=dict(size=11, color=TEXT_DARK)
)

fig.update_yaxes(title_text="Revenue ($M)", row=2, col=1, showgrid=True, gridcolor=GRID)
fig.update_yaxes(title_text="Revenue ($M)", row=2, col=3, showgrid=True, gridcolor=GRID)
fig.update_xaxes(showgrid=False, row=2, col=1)
fig.update_xaxes(showgrid=False, row=2, col=3)
fig.update_xaxes(title_text="Spend ($M)", row=3, col=1, showgrid=True, gridcolor=GRID)
fig.update_yaxes(row=3, col=1, automargin=True, tickfont=dict(size=10))
fig.update_xaxes(title_text="Weekly media spend ($000)", row=3, col=3, showgrid=True, gridcolor=GRID)
fig.update_yaxes(title_text="Revenue ($M)", row=3, col=3, showgrid=True, gridcolor=GRID)

fig.update_layout(
    width=1550,
    height=1150,
    paper_bgcolor="white",
    plot_bgcolor="white",
    margin=dict(l=150, r=70, t=190, b=110),
    title=dict(
        text=(
            "<b>SOLMAR EYEWEAR</b><br>"
            "<span style='font-size:14px;color:#666'>Weekly Performance Dashboard · Aug 2024 – Aug 2026</span>"
        ),
        x=0.02, y=0.998, xanchor="left", yanchor="top"
    ),
    font=dict(family="Arial", size=11, color=TEXT_DARK),
    hovermode="closest",
    legend=dict(orientation="h", x=0.02, y=0.925, yanchor="top")
)

for annot in fig["layout"]["annotations"]:
    if annot["text"] in (
        "Revenue Trend", "Monthly Revenue by Customer Type",
        "Where Media Budget Is Going", "Revenue vs Media Spend"
    ):
        annot["font"] = dict(size=14, color=TEXT_DARK)

fig.update_xaxes(showline=True, linewidth=1, linecolor="lightgray", row=2)
fig.update_xaxes(showline=True, linewidth=1, linecolor="lightgray", row=3)
fig.update_yaxes(showline=False)

fig.add_annotation(
    x=0, y=-0.06, xref="paper", yref="paper",
    text="Source: Solmar Eyewear weekly master data · Revenue / media spend is descriptive, not campaign-attributed ROAS.",
    showarrow=False, xanchor="left", font=dict(size=9, color=TEXT_MUTED)
)

fig.show()
fig.write_html("Solmar_Eyewear_Interactive_Dashboard.html", include_plotlyjs=True)
print("Dashboard saved as: Solmar_Eyewear_Interactive_Dashboard.html")
```
